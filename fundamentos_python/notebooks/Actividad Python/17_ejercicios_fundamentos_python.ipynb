### Pregunta 1.3: Contar frecuencia de elementos

Dada una lista de nombres repetidos, muestra **cuántas veces aparece cada nombre**.
Cada nombre debe mostrarse una sola vez, en el orden en que aparece por primera vez.

```python
nombres = ['Ana', 'Luis', 'Ana', 'Carlos', 'Luis', 'Ana', 'María', 'Carlos']
```

```python
frecuencias = {}

for nombre in nombres:
    if nombre not in frecuencias:
        frecuencias[nombre] = nombres.count(nombre)

for nombre, cantidad in frecuencias.items():
    print(f"{nombre}: aparece {cantidad} veces")
```

### Pregunta 3.1: Consolidación de inventarios

Combina tres inventarios provenientes de distintas sucursales. Cuando un producto aparezca en varias
sucursales, **suma sus cantidades** y calcula también **en cuántas sucursales** está disponible.

**Requisitos:**
- Crea un diccionario final donde cada producto tenga las claves `cantidad_total` y `sucursales`.
- No modifiques los diccionarios originales.
- Ordena el resultado alfabéticamente por producto.

```python
inventario_norte = {"laptop": 4, "mouse": 10, "monitor": 3}
inventario_sur = {"laptop": 2, "teclado": 8, "monitor": 5}
inventario_centro = {"mouse": 6, "teclado": 4, "webcam": 7}
```

```python
inventario_final = {}
todos_los_inv = [inventario_norte, inventario_sur, inventario_centro]

for inv in todos_los_inv:
    for prod, cant in inv.items():
        if prod not in inventario_final:
            inventario_final[prod] = {"cantidad_total": 0, "sucursales": 0}

        inventario_final[prod]["cantidad_total"] += cant
        inventario_final[prod]["sucursales"] += 1

productos_ordenados = list(inventario_final.keys())
productos_ordenados.sort()

for p in productos_ordenados:
    print(f"{p}: {inventario_final[p]}")
```

- En cada grupo almacena `nombres`, `cantidad` y `salario_promedio`.
- La estructura debe ser un diccionario anidado: `{departamento: {nivel: {...}}}`.

```python
empleados = [
    {"nombre": "Ana", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5200},
    {"nombre": "Luis", "departamento": "Diseño", "nivel": "Junior", "salario": 2800},
    {"nombre": "Carla", "departamento": "Desarrollo", "nivel": "Junior", "salario": 3400},
    {"nombre": "Pedro", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5600},
    {"nombre": "Lucía", "departamento": "Diseño", "nivel": "Senior", "salario": 4700},
]
```

### Pregunta 3.4: Sistema de votación con validación

Procesa una lista de votos y separa los votos **válidos** de los **inválidos**.

**Requisitos:**
- Los votos deben ser insensibles a mayúsculas (`"ANA"` cuenta para `"Ana"`).
- Registra el conteo por candidato, los votos inválidos y el porcentaje obtenido (sobre los votos válidos).
- Determina si hay un ganador o un empate.

```python
candidatos = {"Ana", "Luis", "Carla"}
votos = ["Ana", "Luis", "ana", "Carla", "Pedro", "Luis", "ANA", "", "Carla"]
```

```python
votos_candidatos = {}
for c in candidatos:
    votos_candidatos[c.lower()] = 0

validos = 0
invalidos = 0

for v in votos:
    v_limpio = v.strip().lower()
    if v_limpio in votos_candidatos:
        votos_candidatos[v_limpio] += 1
        validos += 1
    else:
        invalidos += 1

print(f"Total válidos: {validos}")
print(f"Total inválidos: {invalidos}")

max_votos = 0
ganadores = []

for c, cantidad in votos_candidatos.items():
    if validos > 0:
        porcentaje = (cantidad / validos) * 100
    else:
        porcentaje = 0
    print(f"{c.title()}: {cantidad} votos ({porcentaje:.2f}%)")

    if cantidad > max_votos:
        max_votos = cantidad
        ganadores = [c.title()]
    elif cantidad == max_votos:
        ganadores.append(c.title())

if len(ganadores) == 1:
    print(f"Ganador: {ganadores[0]}")
elif len(ganadores) > 1:
    print(f"Empate entre: {ganadores}")
```

### Pregunta 3.5: Índice invertido de documentos

Crea un **índice invertido** que permita saber en qué documentos aparece cada palabra.

**Requisitos:**
- Cada palabra debe apuntar a un **conjunto** (`set`) de identificadores de documentos.
- Ignora las palabras de una sola letra.
- Usa el índice para responder: ¿en qué documentos aparecen `"python"` y `"datos"`?

```python
documentos = {
    "doc1": "python y ciencia de datos",
    "doc2": "python para desarrollo web",
    "doc3": "ciencia de datos con inteligencia artificial",
}
```

```python
indice = {}

for doc_id, texto in documentos.items():
    palabras = texto.split()
    for p in palabras:
        if len(p) > 1:
            if p not in indice:
                indice[p] = set()
            indice[p].add(doc_id)

for p, docs in indice.items():
    print(f"{p}: {docs}")

python_docs = indice.get("python", set())
datos_docs = indice.get("datos", set())
ambos = python_docs.intersection(datos_docs)
print(f"Aparecen en ambos: {ambos}")
```

### Pregunta 5.1: Analizar las ventas de una tienda

Crea una función `analizar_ventas(ventas)` que reciba una lista de montos de venta y construya un resumen estadístico.

**Requisitos:**
- Calcular el total vendido, el promedio, la venta más alta y la venta más baja.
- Contar cuántas ventas fueron superiores al promedio.
- Indicar cuántas ventas se encuentran por debajo de 100.
- Retornar todos los resultados dentro de un diccionario.
- Si la lista está vacía o contiene valores negativos, retornar un mensaje de error.

**Reto adicional:** agrega un parámetro opcional `meta`. La función debe indicar si el total vendido alcanzó dicha meta.

```python
ventas = [150.50, 200, 80.90, 300, 175.50, 90]
```

```python
def analizar_ventas(ventas, meta=None):
    if len(ventas) == 0:
        return "Error: la lista está vacía"

    for v in ventas:
        if v < 0:
            return "Error: hay ventas negativas"

    total = sum(ventas)
    promedio = total / len(ventas)
    maxima = max(ventas)
    minima = min(ventas)

    superan_prom = 0
    bajo_100 = 0

    for v in ventas:
        if v > promedio:
            superan_prom += 1
        if v < 100:
            bajo_100 += 1

    resultado = {
        "total": total,
        "promedio": promedio,
        "maxima": maxima,
        "minima": minima,
        "mas_que_promedio": superan_prom,
        "bajo_100": bajo_100
    }

    if meta is not None:
        resultado["alcanzo_meta"] = total >= meta

    return resultado

print(analizar_ventas(ventas, meta=1000))
```

### Pregunta 5.2: Validar una contraseña y explicar los errores

Crea una función `validar_password(password)` que determine si una contraseña cumple todas las reglas de seguridad.

**Requisitos:**
- Debe tener al menos 8 caracteres.
- Debe contener una letra mayúscula, una minúscula y un número.
- Debe incluir al menos uno de estos símbolos: `@`, `#`, `$` o `%`.
- No debe contener espacios.
- Debe retornar una tupla con el formato `(es_valida, errores)`, donde `errores` sea una lista.

```python
password_prueba = 'Python123'
```

```python
def validar_password(password):
    errores = []

    if len(password) < 8:
        errores.append("Muy corta, necesita 8 caracteres")

    tiene_mayus = False
    tiene_minus = False
    tiene_num = False

    for letra in password:
        if letra.isupper():
            tiene_mayus = True
        elif letra.islower():
            tiene_minus = True
        elif letra.isdigit():
            tiene_num = True

    if not tiene_mayus:
        errores.append("Falta mayúscula")
    if not tiene_minus:
        errores.append("Falta minúscula")
    if not tiene_num:
        errores.append("Falta un número")

    simbolos = ["@", "#", "$", "%"]
    tiene_simbolo = False
    for s in simbolos:
        if s in password:
            tiene_simbolo = True
            break

    if not tiene_simbolo:
        errores.append("Falta símbolo (@, #, $, %)")

    if " " in password:
        errores.append("No puede tener espacios")

    es_valida = len(errores) == 0
    return (es_valida, errores)

valida, fallos = validar_password(password_prueba)
if valida:
    print("Contraseña correcta")
else:
    print("Contraseña incorrecta por:")
    for f in fallos:
        print("-", f)
```

- Incluir únicamente a quienes tengan un promedio igual o superior a `minimo`.
- Retornar una lista de tuplas con `(nombre, promedio, clasificación)`.
- Ordenar de mayor a menor promedio y, en caso de empate, alfabéticamente.

```python
empleados = {
    'Ana': [80, 90, 75],
    'Carlos': [60, 70, 65],
    'Lucía': [95, 85, 90],
    'Pedro': [70, 70, 70],
}
```

```python
def evaluar_empleados(empleados, minimo=70):
    resultado = []

    for nombre, notas in empleados.items():
        # Promedio simple
        promedio = round(sum(notas) / len(notas), 2)

        # Clasificación
        if promedio >= 90:
            clasificacion = "Excelente"
        elif promedio >= 80:
            clasificacion = "Bueno"
        elif promedio >= 70:
            clasificacion = "Regular"
        else:
            clasificacion = "Deficiente"

        # Filtrar según el puntaje mínimo
        if promedio >= minimo:
            resultado.append((nombre, promedio, clasificacion))

    # Ordenar: primero por promedio descendente y luego por nombre alfabéticamente
    resultado.sort(key=lambda x: (-x[1], x[0]))
    return resultado

# Probamos con los empleados del ejercicio
empleados_evaluados = evaluar_empleados(empleados, minimo=70)
print("Empleados aprobados:")
for emp in empleados_evaluados:
    print(emp)
```

### Pregunta 5.4: Comprimir y descomprimir texto

Implementa dos funciones: `comprimir_texto(texto)` y `descomprimir_texto(texto)`.

**Requisitos:**
- `comprimir_texto` debe reemplazar caracteres consecutivos repetidos por el carácter y su cantidad.
  Por ejemplo, `aaabbccccdaa` debe convertirse en `a3b2c4d1a2`.
- `descomprimir_texto` debe reconstruir el texto original.
- La descompresión debe admitir cantidades de más de un dígito, como `a12`.
- Si el formato comprimido es inválido, retornar un mensaje de error.

```python
texto = 'aaabbccccdaa'
```

```python
def comprimir_texto(texto):
    if len(texto) == 0:
        return ""

    comprimido = ""
    letra_actual = texto[0]
    contador = 1

    for i in range(1, len(texto)):
        if texto[i] == letra_actual:
            contador += 1
        else:
            comprimido = comprimido + letra_actual + str(contador)
            letra_actual = texto[i]
            contador = 1

    # Agregamos la ultima letra y su cantidad
    comprimido = comprimido + letra_actual + str(contador)
    return comprimido


def descomprimir_texto(texto_comp):
    if len(texto_comp) == 0:
        return ""

    descomprimido = ""
    i = 0
    n = len(texto_comp)

    while i < n:
        letra = texto_comp[i]

        # Debe ser una letra
        if not letra.isalpha():
            return "ERROR: Formato inválido."
        i += 1

        # Extraemos el numero correspondiente
        numero_str = ""
        while i < n and texto_comp[i].isdigit():
            numero_str += texto_comp[i]
            i += 1

        if numero_str == "":
            return "ERROR: Formato inválido."

        cantidad = int(numero_str)
        descomprimido += letra * cantidad

    return descomprimido

#Porbamos
texto_comp = comprimir_texto(texto)
print(f"Texto original:    {texto}")
print(f"Texto comprimido:   {texto_comp}")

texto_desc = descomprimir_texto(texto_comp)
print(f"Texto recuperado:   {texto_desc}")
```

### Pregunta 5.5: Sistema de reservas de asientos

Crea una función `reservar_asientos(sala, asientos_solicitados)` que permita reservar varios asientos de una sala.

**Requisitos:**
- La sala será una lista de listas donde `L` significa libre y `O` significa ocupado.
- Cada asiento solicitado se expresará mediante una tupla `(fila, columna)`, **numeradas desde 1**.
- Validar que todas las posiciones existan y que los asientos estén libres.
- Realizar la reserva únicamente si todos están disponibles.
- Evitar reservas parciales y mantener la sala intacta si ocurre un error.
- Retornar el estado, los asientos reservados y la sala actualizada.

**Reto adicional:** crea otra función `mostrar_sala(sala)` que muestre filas y columnas de manera legible.

```python
sala = [['L', 'L', 'O', 'L'],
        ['O', 'L', 'L', 'L'],
        ['L', 'O', 'L', 'L']]
asientos = [(1, 2), (2, 3), (3, 1)]
```

```python
def mostrar_sala(sala):
    print("Estado actual de la sala:")
    for num_fila, fila in enumerate(sala, start=1):
        print(f"Fila {num_fila}: {fila}")

def reservar_asientos(sala, asientos_solicitados):
    total_filas = len(sala)
    total_columnas = len(sala[0])

    # 1. Validar que todas las posiciones existan y estén libres ('L')
    for f, c in asientos_solicitados:
        idx_f = f - 1
        idx_c = c - 1

        # Fuera de rango
        if idx_f < 0 or idx_f >= total_filas or idx_c < 0 or idx_c >= total_columnas:
            return f"ERROR: El asiento ({f}, {c}) no existe.", [], sala

        # Ya ocupado
        if sala[idx_f][idx_c] == 'O':
            return f"ERROR: El asiento ({f}, {c}) ya está ocupado.", [], sala

    # 2. Si todos están disponibles, creamos una copia limpia de la sala
    sala_nueva = []
    for fila in sala:
        sala_nueva.append(fila.copy())

    # Realizamos la reserva marcando 'O'
    for f, c in asientos_solicitados:
        sala_nueva[f - 1][c - 1] = 'O'

    return "Reserva exitosa", asientos_solicitados, sala_nueva

# Ejecución
mostrar_sala(sala)

mensaje, reservados, sala_actualizada = reservar_asientos(sala, asientos)
print(f"\nResultado: {mensaje}")
print(f"Asientos reservados: {reservados}\n")

mostrar_sala(sala_actualizada)
```

### Pregunta 6.1: Clase CuentaBancaria

Crea una clase `CuentaBancaria` con los atributos `titular` y `saldo` (el saldo inicial por defecto será 0).

- Método `depositar(monto)`: suma al saldo, solo si el monto es positivo.
- Método `retirar(monto)`: resta del saldo, pero debe impedir que el saldo quede negativo y avisar `"Fondos insuficientes"`.
- Método `consultar_saldo()`: devuelve el saldo actual.

Prueba la clase con una cuenta: deposita, retira un monto válido y luego intenta retirar más de lo disponible.

*Objetivo: modificar el estado interno del objeto con validaciones.*

```python
# No hay datos de entrada: crea tus propios objetos para probar la clase.
```

```python
class CuentaBancaria:
    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, monto):
        if monto > 0:
            self.saldo += monto
            print(f"Se depositaron S/. {monto}. Saldo actual: S/. {self.saldo}")
        else:
            print("El monto a depositar debe ser mayor a 0.")

    def retirar(self, monto):
        if monto > self.saldo:
            print("Fondos insuficientes")
        elif monto <= 0:
            print("El monto a retirar debe ser mayor a 0.")
        else:
            self.saldo -= monto
            print(f"Se retiraron S/. {monto}. Saldo actual: S/. {self.saldo}")

    def consultar_saldo(self):
        return self.saldo

# Pruebas de la clase
mi_cuenta = CuentaBancaria("Anthony Mamani", 100)
print(f"Cuenta de: {mi_cuenta.titular} - Saldo inicial: S/. {mi_cuenta.consultar_saldo()}")

mi_cuenta.depositar(50)
mi_cuenta.retirar(70)
mi_cuenta.retirar(200)  # Intento de retirar más de lo disponible
```

### Pregunta 6.4: Composición: Motor dentro de Vehiculo

- Crea una clase `Vehiculo` cuyo constructor reciba una `marca` y un objeto `Motor`.
- El método `arrancar()` de `Vehiculo` debe **delegar** la acción en su motor.
- El motor debe recordar si ya estaba encendido y avisarlo si se intenta arrancar dos veces.

*Objetivo: composición ("tiene un") frente a herencia ("es un").*

```python
# No hay datos de entrada: crea tus propios objetos para probar las clases.
```

```python
class Motor:
    def __init__(self, tipo, caballos):
        self.tipo = tipo
        self.caballos = caballos
        self.encendido = False

    def arrancar(self):
        if self.encendido == True:
            print(f"El motor {self.tipo} ya estaba encendido.")
        else:
            self.encendido = True
            print(f"El motor {self.tipo} de {self.caballos} HP ha arrancado.")

class Vehiculo:
    def __init__(self, marca, motor):
        self.marca = marca
        self.motor = motor

    def arrancar(self):
        print(f"Arrancando vehiculo {self.marca}...")
        # Delegamos la acción en el motor
        self.motor.arrancar()

# Pruebas
motor1 = Motor("Gasolina V6", 250)
mi_camioneta = Vehiculo("Toyota", motor1)

mi_camioneta.arrancar()
mi_camioneta.arrancar()  # Intentar arrancar de nuevo
```

### Pregunta 6.5: Sistema de biblioteca (ejercicio integrador)

Crea tres clases que trabajen juntas: `Libro`, `Usuario` y `Biblioteca`.

- `Libro`: `titulo`, `autor` y `disponible` (`True` por defecto). *(Es una clase distinta a la del ejercicio 6.2: puedes redefinirla.)*
- `Usuario`: `nombre` y una lista de `libros_prestados`.
- `Biblioteca`: guarda una lista de libros y ofrece los métodos `agregar_libro()`, `buscar(titulo)`,
  `prestar(titulo, usuario)`, `devolver(titulo, usuario)` y `catalogo()`.
- Al prestar un libro debe marcarse como no disponible y añadirse a la lista del usuario; al devolverlo, lo contrario.
- Contempla los casos: el libro no existe, el libro ya está prestado, el usuario no tiene ese libro.

*Objetivo: integrar varias clases que se relacionan entre sí, tal como ocurre en un programa real.*

```python
# No hay datos de entrada: crea tus propios objetos para probar las clases.
```

```python
class Libro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor
        self.disponible = True

class Usuario:
    def __init__(self, nombre):
        self.nombre = nombre
        self.libros_prestados = []

class Biblioteca:
    def __init__(self):
        self.libros = []

    def agregar_libro(self, libro):
        self.libros.append(libro)
        print(f"Libro agregado: '{libro.titulo}'")

    def buscar(self, titulo):
        for l in self.libros:
            if l.titulo.lower() == titulo.lower():
                return l
        return None

    def prestar(self, titulo, usuario):
        libro = self.buscar(titulo)

        if libro is None:
            print(f"Error: El libro '{titulo}' no existe en la biblioteca.")
        elif libro.disponible == False:
            print(f"Error: El libro '{titulo}' ya está prestado.")
        else:
            libro.disponible = False
            usuario.libros_prestados.append(libro)
            print(f"Éxito: Se prestó '{titulo}' a {usuario.nombre}.")

    def devolver(self, titulo, usuario):
        # Verificar si el usuario realmente tiene ese libro
        libro_a_devolver = None
        for l in usuario.libros_prestados:
            if l.titulo.lower() == titulo.lower():
                libro_a_devolver = l
                break

        if libro_a_devolver is None:
            print(f"Error: {usuario.nombre} no tiene el libro '{titulo}'.")
        else:
            libro_a_devolver.disponible = True
            usuario.libros_prestados.remove(libro_a_devolver)
            print(f"Éxito: {usuario.nombre} devolvió '{titulo}'.")

    def catalogo(self):
        print("\n--- Catálogo de la Biblioteca ---")
        for l in self.libros:
            estado = "Disponible" if l.disponible else "Prestado"
            print(f"- {l.titulo} ({l.autor}) -> {estado}")
        print("---------------------------------\n")

# Pruebas del sistema
biblioteca = Biblioteca()

l1 = Libro("Cien años de soledad", "Gabriel García Márquez")
l2 = Libro("El principito", "Antoine de Saint-Exupéry")
biblioteca.agregar_libro(l1)
biblioteca.agregar_libro(l2)

usuario1 = Usuario("Anthony")

biblioteca.catalogo()

# Prestar libro
biblioteca.prestar("Cien años de soledad", usuario1)
biblioteca.prestar("Cien años de soledad", usuario1)

biblioteca.catalogo()

# Devolver libro
biblioteca.devolver("Cien años de soledad", usuario1)
biblioteca.catalogo()
```
